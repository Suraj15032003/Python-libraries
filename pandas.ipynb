3. Write a Python program to add, subtract, multiple and divide two Pandas Series.
Sample Series: [2, 4, 6, 8, 10], [1, 3, 5, 7, 9]

```python
series1 = pd.Series([2,4,6,8,10])
series2 = pd.Series([1,3,5,7,9])

print("Series 1:")
print(series1)
print("\nSeries 2:")
print(series2)

addition = series1 + series2
subtraction = series1 - series2
multiplication = series1 * series2
division = series1 / series2

print("\nAddition of Series 1 and Series 2:")
print(addition)

print("\nSubtraction of Series 1 and Series 2:")
print(subtraction)

print("\nMultiplication of Series 1 and Series 2:")
print(multiplication)

print("\nDivision of Series 1 and Series 2:")
print(division)
```

4. Write a Python program to get the powers of an array values element-wise.
Note: First array elements raised to powers from second array
Expected Output:
Original array
[0 1 2 3 4 5 6]

First array elements raised to powers from second array, element-wise:
[ 0 1 8 27 64 125 216]

```python
original_series = pd.Series([0, 1, 2, 3, 4, 5, 6])
print("Original array:")
print(original_series)

powers_series = pd.Series(range(len(original_series)))  # [0, 1, 2, 3, 4, 5, 6]

result = original_series ** powers_series

print("\nFirst array elements raised to powers from second array, element-wise:")
print(result)
```

labels = ['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'i', 'j']
Values for each column will be:
name : "Suresh", score: 15.5, attempts: 1, qualify: "yes", label: "k"

```python
exam_data = {
    'name': ['Anastasia', 'Dima', 'Katherine', 'James', 'Emily', 'Michael', 'Matthew', 'Laura', 'Kevin', 'Jonas'],
    'score': [12.5, 9, 16.5, np.nan, 9, 20, 14.5, np.nan, 8, 19],
    'attempts': [1, 3, 2, 3, 2, 3, 1, 1, 2, 1],
    'qualify': ['yes', 'no', 'yes', 'no', 'no', 'yes', 'yes', 'no', 'no', 'yes']
}

labels = ['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'i', 'j']

df = pd.DataFrame(exam_data, index=labels)

# Display the original DataFrame
print("Original DataFrame:")
print(df)

# Append new row 'k' with specified values using pd.concat
new_row = pd.DataFrame({'name': ['Suresh'], 'score': [15.5], 'attempts': [1], 'qualify': ['yes']}, index=['k'])
df = pd.concat([df, new_row])

# Display the DataFrame with the new row
print("\nDataFrame after appending row 'k':")
print(df)

# Delete the new row 'k' and return to original DataFrame
df = df.drop('k')

# Display the DataFrame after deletion
print("\nOriginal DataFrame after deleting row 'k':")
print(df)
```

22. Write a Python program to iterate over rows in a DataFrame.

```python
exam_data = {
    'name': ['Anastasia', 'Dima', 'Katherine', 'James', 'Emily', 'Michael', 'Matthew', 'Laura', 'Kevin', 'Jonas'],
    'score': [12.5, 9, 16.5, np.nan, 9, 20, 14.5, np.nan, 8, 19],
    'attempts': [1, 3, 2, 3, 2, 3, 1, 1, 2, 1],
    'qualify': ['yes', 'no', 'yes', 'no', 'no', 'yes', 'yes', 'no', 'no', 'yes']
}

labels = ['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'i', 'j']

df = pd.DataFrame(exam_data, index=labels)

# Iterate over rows using iterrows()
print("Iterating over rows in the DataFrame:")
for index, row in df.iterrows():
    print(f"Index: {index}, Row data: {row.to_dict()}")
```
